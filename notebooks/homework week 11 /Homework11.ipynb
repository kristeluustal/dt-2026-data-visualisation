```python
text = "Music is an important part of everyday life. People listen to music when they are happy, sad, relaxed, or excited. There are many different music genres, such as pop, rock, jazz, classical, and hip-hop. Some people enjoy listening to music, while others prefer playing musical instruments or writing their own songs. Music can bring people together, express emotions, and create unforgettable memories. Everyone has their own taste in music, and that is what makes it so special."
words = text.split(" ")
print(words)
#Extract only the fifth word
print(words[4])
#Extract first ten  words
print(words[0:10])
#Extract first letter of the 6th word
print(words[5][0])
#Exctract last letter of the last word
print(words[-1][-1])
#Extract last letter of the second- to last word
print(words[-2][-1])

#number of words in the text
print(len(words))
```

I created my own text about music and used Python to practise working with strings and lists.



```python
#word for word in words means unpacks each separately
print([word for word in words])
#print first letter of every word
print([word[0] for word in words])
#Reverse the entire word list
print(words[::-1])

```

```python
#Print words that have more than five letters
print([word for word in words if len(word) > 5])
```

```python
#import only the class we need
from wordcloud import WordCloud
import matplotlib.pyplot as plt #To draw charts

# Configure dimensions first, then generate wordcloud from text
wc = WordCloud(width=400, height=200, background_color="pink").generate("Music is an important part of everyday life. People listen to music when they are happy, sad, relaxed, or excited. There are many different music genres, such as pop, rock, jazz, classical, and hip-hop. Some people enjoy listening to music, while others prefer playing musical instruments or writing their own songs. Music can bring people together, express emotions, and create unforgettable memories. Everyone has their own taste in music, and that is what makes it so special.")

plt.imshow(wc, interpolation="bilinear")

plt.show()
```

I used a stopwords list to remove common English words that do not carry much meaning, such as "the", "and" and "to". This makes the important words easier to notice.

```python
WordCloud(
    background_color="white",
    stopwords=["the", "and", "to", "of", "a", "is", "in", "but", "because", "also",
               "some", "every", "many", "their", "with", "for", "can", "have"]
).generate_from_text(text).to_image()
```

I used a heart-shaped image for my word cloud. The NumPy library converts the image into an array that WordCloud can use. The words are placed inside the shape, creating a heart-shaped visualisation.

```python
from PIL import Image

Image.open("heart.png")
```
