```python
import pandas as pd
df = pd.read_csv("Homework5data.csv")
df.head() # Show the first few rows
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="author", y="number of poems", legend=False, color="#FFBCC1")
plt.title("Number of poems by an author")
plt.ylabel("Number of poems")
plt.show()
```

This charts shows how many poems by each author there is in my dataset. I notice that Hando Runnal and Anna Haava have the most poems while Julius Oengu has significally less than all the others. 

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="poetry book title", y="number of poems", legend=False, color="#FFBCC1")
plt.title("Number of poems by book")
plt.ylabel("Number of poems")
plt.show()
```

This charts shows how many poems there is in each poetry book. "Kiikajon and kaalepuu" has more than 100 while "Tahkuna" has less than 20. It shows how much the number of poems varies between Estonian poetry books.
 

```python
plt.scatter(df["number of poems"], df["release year"], s=df ["number of poems"]*60, color="#BF242F", alpha=0.6)
#s=df["number of poems"]*60 this means if later release year has higher number on poems their bub is bigger
#alpha=0.6 makes the bubles more transparent
plt.xlabel ("number of poems")
plt.ylabel ("relese year")
plt.title("Release year vs number of poems")
```

This chart shows the relationship between the release year and the number of poems. It can be seen that my dataset contains the most poetry books published between 1960 and 2020. The dataset contains the fewest poetry books published before 1900.


```python
books_by_year = df.groupby("release year")["poetry book title"].count()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(books_by_year.index, books_by_year.values, marker="o")
plt.title("Number of poetry Books by release Year")
plt.xlabel("Release year")
plt.ylabel("Number of poetry books")
plt.grid(True)
plt.show()
```

This chart shows how many poetry books were published in different years. In my dataset, only the 1970s have more than one book published in the same year.


```python
author_counts = df["author"].value_counts().head(5)

plt.figure(figsize=(8, 8))

plt.pie(author_counts.values, labels=author_counts.index, autopct="%1.1f%%")

plt.title("Most frequent authors")

plt.show()
```

This chart shows how many poetry books were published in different years. In my dataset, only the 1970s have more than one book published in the same year.
