```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competencies.csv")
df.head() # Show the first few rows
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.title("Amount of effort")
plt.ylabel("Hours of effort")
plt.show()
```

```python
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
colors=plt.cm.plasma(range(len(avg_hours)))
avg_hours.plot(kind="bar", color="#BF242F")
plt.title("Average amount of effort")
plt.ylabel("Hours")
plt.show()

```

```python
plt.scatter(df["Grade"], df["Hours_Studied"], s=df ["Hours_Studied"]*60, color="#BF242F", alpha=0.6)
#s=df["Hours_Studied"]*60 this means if more hours studied has better grades their bub is bigger
#alpha=0.6 makes the bubles more transparent
plt.xlabel ("Grade")
plt.ylabel ("Hours studied")
plt.title("Grade vs Hors studied")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, y="Hours_Studied")
plt.title("Distribution of Grades vs hours")
plt.show()
plt.figure(figsize=(8,5))
df[["Hours_Studied", "Grades"]].boxplot()
plt.title("Box plot for hours studied and grades")
plt.show
```

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot for hours studied and grades")
plt.show
```
